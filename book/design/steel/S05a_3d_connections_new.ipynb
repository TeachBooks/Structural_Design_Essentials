# 3D Connections New

Overview of the steel connections as small interactive 3D models. The details follow the drawings on the [connections](S04_connections.ipynb) page.

Drag to rotate, right-drag to pan and scroll to zoom. Use *Open full size* for a large view with the layer menu (for example to switch bolts or welds on and off).
