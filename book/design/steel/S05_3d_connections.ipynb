# 3D Connections

Explore the structural connections of steel through interactive 3D visualisations. The details follow the drawings on the [connections](S04_connections.ipynb) page.

Drag with the left mouse button to rotate, with the right mouse button to pan, and scroll to zoom. Use the menu at the top left of each model to switch layers (for example bolts or welds) on and off.
