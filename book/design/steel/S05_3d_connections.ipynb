# 3D Connections

Explore the structural connections of steel through interactive 3D visualisations. The details follow the drawings on the [connections](S04_connections.ipynb) page.

Click a thumbnail to open the 3D viewer in a new tab. In the viewer you can rotate, pan and zoom with the mouse, and use the menu at the top left to switch layers (for example bolts or welds) on and off.
